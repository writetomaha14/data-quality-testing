```python
# ============================================================
# Day 28: Performance Basics - deferred, with one confirmed finding
# ============================================================
# Skipped full hands-on practice, but confirmed one real thing first:
# .cache() / .persist() / .unpersist() throw an exception on Free Edition
# serverless compute. Databricks docs confirm this is a documented,
# intentional limitation - serverless caching APIs are unsupported
# entirely, not just this environment being unusual.
#
# Databricks' stated reasoning: serverless is meant to optimize caching
# automatically, so manual cache() shouldn't be needed. (Worth noting:
# this is genuinely debated among users - some report seeing
# recomputation happen anyway.)
#
# Recommended workaround: write intermediate DataFrames to a Delta
# table (Day 13 skill) instead of caching in memory, when reuse is
# genuinely needed.
#
# Rest of Day 28 (shuffle awareness, filter-early habit) still applies
# conceptually - only the caching section needed this correction.
# ============================================================
```

```python
#Create a DataFrame and write the "slow" version deliberately
from pyspark.sql.functions import col, count, when

data = [(1, "Alice", "IT", 55000), (2, "Ben", "HR", 48000),
        (3, "Cara", None, 62000), (4, "Dev", "Finance", None)]
employees = spark.createDataFrame(data, ["id", "name", "department", "salary"])

# Slow version - no caching, multiple actions
print(employees.count())
print(employees.filter(col("salary") > 50000).count())
employees.display()
```

```python
# Note: .cache() is not supported on Serverless - Spark Connect handles optimization automatically
# On classic clusters, you would use: employees.cache()
print(employees.count())
print(employees.filter(col("salary") > 50000).count())
employees.display()
```
