```python

# Re-use Day 12 volumne path and recreate small data frame
volume_path = "/Volumes/workspace/default/day12_files"

data = [(1, "Alice", "IT", 55000),
        (2, "Ben", "HR", 48000),
        (3, "Cara", "IT", 62000)]
columns = ["id", "name", "department", "salary"]

employees = spark.createDataFrame(data, columns)
employees.display()
```

```python
# Write it as Delta
employees.write.format("delta").mode("overwrite").save(f"{volume_path}/employees_delta")
```

```python
#Read it back and confirm
check = spark.read.format("delta").load(f"{volume_path}/employees_delta")
check.display()
```

```python
# Append some more data - confirm it adds, not replaces
new_batch = spark.createDataFrame([(4, "Dev", "Finance", 45000)], columns)
new_batch.write.format("delta").mode("append").save(f"{volume_path}/employees_delta")
check = spark.read.format("delta").load(f"{volume_path}/employees_delta")
check.display()
```

```python
# Register it as a named table
employees.write.format("delta").mode("overwrite").saveAsTable("workspace.default.employees_delta_table")
spark.sql("SELECT * FROM employees_delta_table").display()
```

```python
# Delete the table at the end
spark.sql("DROP TABLE IF EXISTS workspace.default.employees_delta_table")
print("✓ Table dropped successfully")
```

```python
# Delete the actual Delta files from the Volumne
import shutil
shutil.rmtree("/Volumes/workspace/default/day12_files/employees_delta", ignore_errors=True)
```
