```python
data = [(1, 101, "2025-01-10", 500),
        (2, 101, "2025-02-15", 700),
        (3, 101, "2025-03-01", 700),
        (4, 102, "2025-01-20", 300),
        (5, 102, "2025-01-20", 300)]
columns = ["order_id", "customer_id", "order_date", "amount"]

orders = spark.createDataFrame(data, columns)
orders.display()

# Note: orders.rdd.getNumPartitions() is not available on Serverless
# Partition count is managed automatically by Spark on serverless compute
```

```python
# 1. Repartition: Increases or decreases partitions with full shuffle
repartitioned = orders.repartition(4)  # Force 4 partitions
print("✓ Repartitioned to 4 partitions")
print(f"  Data is redistributed across 4 partitions with full shuffle\n")

# 2. Repartition by column: Data with same key goes to same partition
repartitioned_by_col = orders.repartition("customer_id")
print("✓ Repartitioned by customer_id")
print(f"  All orders for customer 101 → same partition")
print(f"  All orders for customer 102 → same partition")
print(f"  Great for: joins, groupBy on customer_id\n")

# 3. Coalesce: Only decreases partitions, no full shuffle (more efficient)
coalesced = orders.coalesce(2)  # Reduce to 2 partitions
print("✓ Coalesced to 2 partitions")
print(f"  Merges existing partitions without full shuffle")
print(f"  Use after: filters that reduce data size\n")

# Key differences:
print("📊 Summary:")
print("  repartition(n)   → Can increase/decrease, full shuffle")
print("  repartition(col) → Partition by key, full shuffle")
print("  coalesce(n)      → Only decrease, no full shuffle (faster)")
```
