```python
from pyspark.sql.functions import col

data = [(1, "Alice", "IT", 55000),
        (2, "Ben", "HR", 48000),
        (3, "Cara", "IT", 62000),
        (4, "Dev", "Finance", 45000)]
columns = ["id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)
df = df.withColumn("salary", col("salary").cast("double"))


# Note: .cache() is not available on Serverless compute
# Serverless manages caching automatically - no manual cache() needed!

df.count()  # This triggers computation
df.display()

# This reuses the DataFrame - Serverless automatically caches hot data
df.filter(col("salary") > 50000).count()
```

```python
# DataFrame Reuse Example - Serverless handles caching automatically

# Perform multiple operations on the same DataFrame
print("Operation 1: Count all employees")
total = df.count()
print(f"Total employees: {total}\n")

print("Operation 2: Filter by department")
it_dept = df.filter(col("department") == "IT")
print(f"IT department: {it_dept.count()} employees\n")

print("Operation 3: Calculate average salary")
avg_salary = df.agg({"salary": "avg"}).collect()[0][0]
print(f"Average salary: ${avg_salary:,.2f}\n")

print("Operation 4: Find high earners")
high_earners = df.filter(col("salary") > 50000)
print(f"High earners (>$50k): {high_earners.count()} employees")
high_earners.display()

print("\n✅ All 4 operations reused the same DataFrame 'df'")
print("✅ Serverless automatically cached hot data - no manual .cache() needed!")
```
